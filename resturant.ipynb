```python
import panel as pn

pn.extension()

panels = []

conversation = []

input_box = pn.widgets.TextInput(
    placeholder="Ask about our menu...",
    width=600
)

button = pn.widgets.Button(
    name="Ask Restaurant Assistant",
    button_type="primary"
)


def ask_restaurant(event):

    question = input_box.value.strip()

    if not question:
        return

    input_box.value = ""

    response, conversation[:] = utils.process_restaurant_message(
        question,
        conversation,
        debug=False
    )

    panels.append(
        pn.Row(
            pn.pane.Markdown(
                f"**Customer:** {question}"
            )
        )
    )

    panels.append(
        pn.Row(
            pn.pane.Markdown(
                f"**Restaurant Assistant:** {response}"
            )
        )
    )

    chat_area.objects = panels


button.on_click(ask_restaurant)

chat_area = pn.Column(
    height=400,
    scroll=True
)

dashboard = pn.Column(
    "# 🍽️ Restaurant AI Assistant",
    input_box,
    button,
    chat_area
)

dashboard
```
